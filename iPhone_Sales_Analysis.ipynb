# Data Analysis and Visualization

# What are the top 10 highest-rated iPhones on Flipkart in India?

```python
df.head()
```

```python
#Top 10 Hightest iphone product name by rating in india
```

```python
highest_rated = df.sort_values(by = ["Star Rating"], ascending = False)
highest_rated = highest_rated.head(10)
```

```python
print(highest_rated["Product Name"])
```

```python
print(highest_rated)
```

# How many ratings do the highest-rated iPhones on Flipkart have?

```python
iphones = highest_rated["Product Name"].value_counts()
label = iphones.index
counts = highest_rated["Number Of Ratings"]
figure = px.bar(highest_rated, x = label, y = counts, title = "Highest No Of Rated iphones")
figure.show()
```

# Which iPhone has the highest number of reviews on Flipkart?

```python
iphones = highest_rated["Product Name"].value_counts()
label = iphones.index
counts = highest_rated["Number Of Reviews"]
figure = px.bar(highest_rated, x = label, y = counts, title = "Highest No Of Reviewed iphones")
figure.show()
```

# What is the relationship between the sale price of iPhones and the number of ratings on Flipkart?

```python
figure1 = px.scatter(data_frame = df, x = "Number Of Ratings", y = "Sale Price", size = "Discount Percentage", 
                     trendline = "ols", title = "Relationship Between Sales price And Number Of Rating")
figure1.show()
```

# What is the relationship between the discount percentage and the number of ratings of iPhones on Flipkart?

```python
figure2 = px.scatter(data_frame = df, x = "Number Of Ratings", y = "Discount Percentage", size = "Sale Price", 
                     trendline = "ols", title = "Relationship Between Discount Percentage And Number Of Rating")
figure2.show()
```

# Can you figure out the least expensive and most expensive iPhones in the Indian market, along with all their specifications?

```python
most_expensive = df.loc[df["Sale Price"].idxmax()]
least_expensive = df.loc[df["Sale Price"].idxmin()]

#Display the results
print("Most EXPENSIVE Product:")
print(most_expensive)

print("\nLeast EXPENSIVE Product:")
print(least_expensive)
```
